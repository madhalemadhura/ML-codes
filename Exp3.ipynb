```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#Dataset
df = pd.read_csv("/content/sal_data.csv")
print(df)

#input and output
X = df["Experience(X)"].values
y = df["Salary(Y)"].values

# Initialize weights
w0 = 4
w1 = 5
learning_rate = 0.01
epochs = 1000
m = len(X)

# Gradient Descent
for i in range(epochs):

    #Prediction
    y_pred = w0 + w1 * X

    #Cost Function
    cost = (1/m) * np.sum((y - y_pred) ** 2)

    #Gradient Calculation
    dw0 = (-2/m) * np.sum(y - y_pred)
    dw1 = (-2/m) * np.sum(X * (y - y_pred))

    #Update Weight
    w0 = w0 - learning_rate * dw0
    w1 = w1 - learning_rate * dw1

# Final prediction
y_pred = w0 + w1 * X

print("final w0 =", w0)
print("final w1 =", w1)

print("\nFinal Cost Function:")
print(cost)

print("\nPredicted Values:")
print(y_pred)

#regression line
plt.scatter(X, y, color="blue", label="Actual Data")
plt.plot(X, y_pred, color="red", label="Regression Line")

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Linear Regression using Gradient Descent")

plt.legend()
plt.show()
```
