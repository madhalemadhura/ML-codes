```python

import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv("/content/water_potability-selected-columns.csv")

print("Dataset:")
print(df.head())

# Handle Missing Values
df = df.fillna(df.median(numeric_only=True))

# Separate Features and Target
X = df.drop("Potability", axis=1)
y = df["Potability"]

# Find K using sqrt(N)
N = len(df)

sqrt_k = np.sqrt(N)
K = int(sqrt_k)

# Make K odd
if K % 2 == 0:
    K = K + 1

print("\nNumber of data points (N):", N)
print("Square Root of N:", sqrt_k)
print("Selected K:", K)


# Calculate Manhattan Distance
def manhattan_distance(x1, x2):
    return np.sum(np.abs(x1 - x2))


# KNN Prediction
predictions = []
averages = []

for i in range(len(X)):

    distances = []

    current_point = X.iloc[i].values

    # Calculate Manhattan distance
    for j in range(len(X)):

        if i == j:
            continue

        other_point = X.iloc[j].values

        distance = manhattan_distance(
            current_point,
            other_point
        )

        distances.append((distance, y.iloc[j]))

    # Sort distances
    distances.sort(key=lambda x: x[0])

    # Select K nearest neighbors
    nearest_neighbors = distances[:K]

    # Get target values
    neighbor_values = [
        value for distance, value in nearest_neighbors
    ]

    # Calculate average
    average = np.mean(neighbor_values)
    averages.append(average)

    # Prediction
    if average >= 0.5:
        prediction = 1
    else:
        prediction = 0

    predictions.append(prediction)


# -----------------------------------------
# Display only first 10 predictions
# -----------------------------------------

print("\nFirst 10 Predictions:")
print("--------------------------------")

for i in range(10):
    print(
        "Data Point:", i,
        "| K:", K,
        "| Average:", round(averages[i], 4),
        "| Actual:", y.iloc[i],
        "| Prediction:", predictions[i]
    )

```
